# nadayantra

> Shabdayantra on the web: teach sounds to type glyphs, recognised by nada.


A [FastHTML](https://fastht.ml) app. You teach it a few sounds, bind each one to a glyph or key, and it types for you. Sounds are recognised in-process by [nada](https://github.com/alix559/nada) (wav2vec 2.0 on Modular MAX), with an [Observable Framework](https://observablehq.com/framework/) frontend in `frontend/`.


## Install

```sh
pixi install
```

`nada` is expected next to this repo (`../nada`), with its checkpoint in `../nada/models`. Set `NADA_MODELS` to use a different directory.


## Run

```sh
pixi run serve
```

The app listens on <http://localhost:5001>.


## Develop

This is an [nbdev](https://nbdev.fast.ai) project. Library code lives in notebooks under `nbs/` and is exported to the `nadayantra` package: `core`, `engine` (the nada model), `keys` (macOS keyboard access), `api` (JSON for the frontend) and `main` (the FastHTML app). Edit the notebooks, never the exported `.py` files:

```sh
pixi run nbdev-export
pixi run nbdev-test
```
