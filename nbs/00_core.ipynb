```python
#| export
from pathlib import Path
```

```python
#| export
def repo_root() -> Path:
    "Checkout that holds `pyproject.toml`, `glyph_table.json`, `static/` and `models/`."
    starts = []
    file = globals().get("__file__")
    if file: starts.append(Path(file).resolve().parent)
    starts.append(Path.cwd())
    for start in starts:
        for parent in (start, *start.parents):
            if (parent/"pyproject.toml").exists() and (parent/"glyph_table.json").exists(): return parent
    raise FileNotFoundError("Could not find the nadayantra checkout (no pyproject.toml next to glyph_table.json).")
```
